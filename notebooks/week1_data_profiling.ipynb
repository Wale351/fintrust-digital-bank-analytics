## Week 1 Profiling Summary

My initial review shows that the FinTrust datasets are generally well structured for the planned analysis.

The customer dataset contains 1,500 unique customer records with no missing values or duplicate customer IDs.

The transaction dataset contains 12,000 unique transaction records. A small amount of missing information was identified in Device_Type and Location.

All transaction records successfully link to a valid customer using Customer_ID, confirming a one-to-many relationship between customers and transactions.
